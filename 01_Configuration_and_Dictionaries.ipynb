**Input:** none
**Output:** `Dictionary/column_renames.json`, `status_mapping.json`, `element_mapping.json`, `crossborder_code.json`

Defines the mappings used to harmonise the nine TYNDP editions, whose column
headers and status labels change from cycle to cycle. Over 100 source headers
are mapped to a common schema, and textual progress labels to the ordinal
lifecycle scale (1 Consideration, 2 Planned, 3 Design & Permitting,
4 Construction, 5 Commissioned, 6 Cancelled).
