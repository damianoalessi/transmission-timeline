# A_3b — Final investment outcomes

**Input:** `Results/07.xlsx`
**Output:** counts of commissioned, cancelled and ongoing investments

Reports the terminal status of the tracked portfolio: the baseline attrition
figures used throughout the paper.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Results/07.xlsx", header=[0,1], index_col=0)
```

```python
conclusion_col = df[('meta', 'Investment Conclusion')]

conclusion_stats = conclusion_col.value_counts(dropna=False)

n_commissioned = conclusion_stats.get('commissioned', 0)
n_cancelled = conclusion_stats.get('cancelled', 0)
n_empty = conclusion_col.isna().sum()

print("\n" + "="*40)
print(" INVESTMENT CONCLUSION SUMMARY")
print("="*40)
print(f"Commissioned:  {n_commissioned}")
print(f"Cancelled:     {n_cancelled}")
print(f"Empty (NaN):   {n_empty}")
print("-" * 40)
print("All unique values found:")
print(conclusion_stats)
print("="*40 + "\n")
```
