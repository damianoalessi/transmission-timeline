# A_4_4 — Cancellation drivers

**Input:** `Data/Project non submitted-non compliant_2020_2026.xlsx`
**Output:** `Results/Cluster_Cancelled.xlsx` (SI Table 13)

The 51 projects removed from the 2020–2026 cycles carry short, formulaic
rationales, so a keyword classifier replaces the embedding pipeline used for the
longer delay statements. Every assignment was subsequently checked against the
original text; the script is provided as a replicability aid rather than as the
final classifier.

```python
import pandas as pd
```

```python
df_cancelled = pd.read_excel("Data/Project non submitted-non compliant_2020_2026.xlsx")
```

```python
# Define the categorization function
def categorize_cancellation(text):
    text = str(text).lower().strip()
    
    # 1. NO RATIONALE
    if text in ["no information", "cancelled", "nan", "", "-"]:
        return "NO_RATIONALE_PROVIDED"
        
    # 2. ADMINISTRATIVE & COMPLIANCE
    elif any(keyword in text for keyword in ["deadline", "criteria", "ceased to exist", "eligible", "pci", "pmi"]):
        return "ADMINISTRATIVE_AND_COMPLIANCE_FAILURE"
        
    # 3. RECLUSTERING
    elif any(keyword in text for keyword in ["reclustering", "replaced"]):
        return "RECLUSTERING"
        
    # 4. PERMITTING & SITING
    elif any(keyword in text for keyword in ["permits", "environmental", "social impact", "siting"]):
        return "PERMITTING_AND_SITING_ROADBLOCKS"
        
    # 5. WITHDRAWN & HOLD
    elif any(keyword in text for keyword in ["withdrawn by project promoter", "on hold", "internal reasons"]):
        return "WITHDRAWN_OR_ON_HOLD"
        
    # 6. STRATEGIC OBSOLESCENCE
    elif any(keyword in text for keyword in ["no need", "policy", "not considered necessary", "no longer needed", "support", "cross-border"]):
        return "STRATEGIC_OBSOLESCENCE_AND_POLICY"
        
    # Fallback
    else:
        return "UNCLASSIFIED"


df_cancelled['Expert_Category'] = df_cancelled['Reason_of_cancellation'].apply(categorize_cancellation)
df_cancelled.to_excel("Results/Cluster_Cancelled.xlsx", index=False)
```
