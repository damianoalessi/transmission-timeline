# A_5 — Category shares

**Input:** `0_Clustering_Manual_Revision/Delayed_Reviewed.xlsx`, `Rescheduled_Reviewed.xlsx`, `Cancelled_Reviewed.xlsx`
**Output:** Table 2, SI Tables 11–13

Aggregates the manually reviewed labels into the shares reported in the paper.
The ten delay categories are collapsed into five macro-categories;
reschedulings are reported at their native five-category level. Percentages are
computed over the full set of flagged statements, including those carrying no
justification.

```python
import pandas as pd
```

```python
df_rescheduled_final = pd.read_excel("0_Clustering_Manual_Revision/Rescheduled_Reviewed.xlsx")
df_delayed_final = pd.read_excel("0_Clustering_Manual_Revision/Delayed_Reviewed.xlsx")
df_cancelled_final = pd.read_excel("0_Clustering_Manual_Revision/Cancelled_Reviewed.xlsx")
```

104 Rows were empty and had no explanation for Rescheduled

8 Rows were empty and had no explanation for Delayed

## Rescheduled

```python
category_counts = df_rescheduled_final['Expert_Category'].value_counts()

if 'NO_CHANGE_OR_DATA_ISSUE' in category_counts:
    category_counts['NO_CHANGE_OR_DATA_ISSUE'] += 104
else:
    category_counts['NO_CHANGE_OR_DATA_ISSUE'] = 104

total_count = category_counts.sum()
category_percentage = (category_counts / total_count) * 100

summary_df = pd.DataFrame({
    'Count': category_counts, 
    'Percentage': category_percentage.map("{:.1f}%".format)
})

print("--- Definitive Sentence Count per Category (Including 80 'NO_CHANGE' entries) ---\n")
print(summary_df)
```

## Deleyed

```python
macro_mapping = {
    'GENERAL_BUREAUCRACY': 'Regulatory & Permitting Framework',
    'LEGAL_AND_REGULATORY': 'Regulatory & Permitting Framework',
    'ENV_PERMIT_EIA': 'Regulatory & Permitting Framework',
    
    'CASCADING_DELAYS': 'Strategic & Systemic Interdependencies',
    'SOCIAL_AND_ROUTING': 'Strategic & Systemic Interdependencies',
    
    'FUNDING_AND_FINANCE': 'Financial & Industrial Execution',
    'PROCUREMENT_AND_SUPPLY_CHAIN': 'Financial & Industrial Execution',
    'LAND_ACQUISITION': 'Financial & Industrial Execution',
    
    'SCOPE_AND_DESIGN_CHANGES': 'Technical Scope & Data',
    
    'NO_DELAY_OR_DATA_ISSUE': 'Data Issue / No Delay',
    'NO_CHANGE_OR_DATA_ISSUE': 'Data Issue / No Delay',
    'UNCLASSIFIED': 'Data Issue / No Delay'
}

df_delayed_final['Macro_Category'] = df_delayed_final['Expert_Category'].map(macro_mapping)

category_counts = df_delayed_final['Macro_Category'].value_counts()

if 'Data Issue / No Delay' in category_counts:
    category_counts['Data Issue / No Delay'] += 5
else:
    category_counts['Data Issue / No Delay'] = 5

total_count = category_counts.sum()
category_percentage = (category_counts / total_count) * 100

summary_df = pd.DataFrame({
    'Count': category_counts, 
    'Percentage': category_percentage.map("{:.1f}%".format)
})

print("--- Definitive Sentence Count per Macro-Category ---\n")
print(summary_df)
```

## Cancelled

```python
category_counts = df_cancelled_final['Expert_Category'].value_counts()

total_count = category_counts.sum()
category_percentage = (category_counts / total_count) * 100

summary_df = pd.DataFrame({
    'Count': category_counts, 
    'Percentage': category_percentage.map("{:.1f}%".format)
})

print("--- Definitive Sentence Count per Category (Including 80 'NO_CHANGE' entries) ---\n")
print(summary_df)
```
